```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential 
from tensorflow.keras.models import load_model
```

```python
## LOAD THE IMDB DATASET WORD INDEX

word_index= imdb.get_word_index()
reverse_word_index={value:key for key, value in word_index.items()}
```

```python
## LOAD THE PRETRAINED MODEL

model=load_model('simple_rnn_imdb.h5')
model.summary()
```

```python
model.get_weights()
```

```python
# Step 2: Helper Functions

# Function to decode reviews
def decode_review(encoded_review):
    return ' '.join([reverse_word_index.get(i - 3, '?') for i in encoded_review])

# Function to preprocess user input
def preprocess_text(text):

    ## Convert the input text to lowercase and split it into individual words
    words = text.lower().split()

                ## Convert each word into its integer index
               ## Loop through each word in 'words'
               ## word_index.get(word, 2):
               ##    - If the word exists, return its index
               ##    - If the word is not found, return 2 (<UNK> = Unknown word)
               ## Add 3 because the IMDB dataset reserves the first 3 indices for special tokens
               ## (<PAD>, <START>, <UNK>)
    encoded_review = [word_index.get(word, 2) + 3 for word in words]

               ## Pad the review with 0s so its length becomes 500
               ## This ensures every input has the same length required by the model
    padded_review = sequence.pad_sequences([encoded_review], maxlen=500)

               ## Return the preprocessed review
    return padded_review
```

## PREDICTION FUNCTION

```python
def predict_sentiment(review):
    preprocessed_input=preprocess_text(review)

    prediction=model.predict(preprocessed_input)

    sentiment= 'Positive' if prediction[0][0] > 0.5 else 'Negative'

    return sentiment,prediction
```

```python
# Step 4: User Input and Prediction

# Example review for prediction
example_review = "This movie was fantastic! The acting was great and the plot was thrilling."

sentiment, score = predict_sentiment(example_review)

print(f'Review: {example_review}')
print(f'Sentiment: {sentiment}')
print(f'Prediction Score: {score}')
```
