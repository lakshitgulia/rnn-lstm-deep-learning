```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good',
]
```

```python
## DEFINE VOCAB SIZE
voc_size=10000
```

```python
## ONE HOT REPRESENTATION

one_hot_repr=[one_hot(words,voc_size)for words in sent]
one_hot_repr

## Apply one_hot() encoding to every sentence in the list 'sent'
## 'for words in sent' loops through each sentence one by one and stores it temporarily in the variable 'words'
## one_hot() converts each word in the current sentence into a unique integer index
## The integers are generated based on the vocabulary size (voc_size)
## The result is stored as a list of lists in 'one_hot_repr'
```

```python
## WORD EMBEDDING REPRESENTATION

from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential

 
```

```python
import numpy as np
```

```python
sent_length = 8

embedded_docs = pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
embedded_docs

## Set the maximum sentence length to 8
## Add 0s at the beginning ('pre') if a sentence is shorter than 8
## Store the padded sentences in 'embedded_docs'
```

```python
## FEATURE REPRESENTATION
dim=10 ## 10 dimensions (10 features)
```

```python
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')

## Create a Sequential neural network model
## Add an Embedding layer to convert integer indices into dense vectors
## voc_size = vocabulary size, dim = embedding vector size, input_length = sentence length
## Compile the model using Adam optimizer and Mean Squared Error (MSE) loss

```

```python
model.summary()
```

```python
model.predict(embedded_docs )

## Pass the padded sentences through the Embedding layer
## Generate the embedding vector for each word in every sentence
## Return the output embeddings
```
